```python
import pickle
from sklearn.model_selection import train_test_split
from sklearn import metrics, svm
import os
import numpy as np
from skimage.io import imread
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from skimage.transform import resize
from sklearn.ensemble import HistGradientBoostingClassifier
```

```python
image_directory = 'C:/Users/Mann/team-60-project/images/images'
batch_size = 5000
```

```python
def batch_load(image_directory):
    x = []
    y = []
    count = 0

    for name in os.listdir(image_directory):
        path = os.path.join(image_directory, name)
        if os.path.isdir(path):
            for image in os.listdir(path):
                image_path = os.path.join(path, image)
                img = imread(image_path)
                resized = resize(img, (128, 128, 3))
                x.append(resized.astype(np.float16).flatten())  
                y.append(name)
                count += 1
                if count % batch_size == 0:
                    yield np.array(x), np.array(y)
                    x = []
                    y = []
    if x:
        yield np.array(x), np.array(y)
```

# HistGradientBoostingClassifier

Utilized the HistGradientBoostingClassifier model which took around ~1209 minutes to train, but did not produce great results. When tested the model gave outputs of "waffles" for almost all images that it was tested upon leading us to look towards another model to train.

```python
model = HistGradientBoostingClassifier()
```

```python
for x, y in batch_load(image_directory):
    x = np.array(x)
    y = np.array(y).flatten()
    X_train, X_test, y_train, y_test = train_test_split(x, y, train_size=0.8, random_state=42)
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    acc = metrics.accuracy_score(y_test, pred)
```
